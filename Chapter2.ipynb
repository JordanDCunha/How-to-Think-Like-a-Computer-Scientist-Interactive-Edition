# Chapter 2.2: Values and Data Types
A **value** is one of the fundamental things—like a word or a number—that a program manipulates. We often refer to these values as **objects**.

### Data Types:
* **int**: Integers (whole numbers like 4, 17, -5).
* **float**: Floating-point numbers (decimals like 3.2, 0.5).
* **str**: Strings (text wrapped in quotes like "Hello").

````python
# Activity 2.2.1 - 2.2.3: Checking Types
print(type("Hello, World!"))
print(type(17))
print(type(3.2))

# What happens if a number is in quotes?
print(type("17"))
print(type("3.2"))
````

### Quotation Marks in Strings
Python supports three ways to define strings:
1.  **Single quotes**: `'Hello'`
2.  **Double quotes**: `"Hello"`
3.  **Triple quotes**: `"""Hello"""` (Used for multi-line strings).

**The Comma Rule:**
In Python, `42,000` is not forty-two thousand. The comma acts as a separator, creating a collection of two different integers (42 and 0).

````python
# Activity 2.2.5 - 2.2.9: Quotes and Commas
print('''"Oh no", she exclaimed, "Ben's bike is broken!"''')

print("""This message will span
several lines
of the text.""")

print(42000)   # One integer
print(42, 000) # Two values separated by a space
````

# Chapter 2.3: Type Conversion Functions

Sometimes you need to change a value from one type to another (e.g., turning a string "100" into a number you can add). Python provides built-in **type conversion functions** to do this:

* `int()`: Converts a value to an integer.
* `float()`: Converts a value to a floating-point number.
* `str()`: Converts a value to a string.

### Important: Truncation
When you convert a `float` to an `int`, Python does **not** round to the nearest number. Instead, it performs **truncation**, which means it simply chops off everything after the decimal point.

````python
# Activity 2.3.1: Converting to Integer
print("--- Truncation Examples ---")
print(3.14, int(3.14))
print(3.9999, int(3.9999))       # Note: This does NOT round to 4!
print(-3.999, int(-3.999))       # Result is -3 (chops off the decimal)

print("\n--- String to Int ---")
print("2345", int("2345"))       # Successfully parses string to int

# UNCOMMENT THE LINE BELOW TO SEE A RUNTIME ERROR
# print(int("23bottles"))        # Error: The string must be a legal number
````

### Converting to Float and String

* **float()**: Can turn an integer or a numeric string into a decimal.
* **str()**: Can turn any value into a string representation. Even though the quotes aren't visible when you `print()` a string, the `type()` function will confirm it has changed.

````python
# Activity 2.3.2 & 2.3.3: Float and String Conversion

print("--- Converting to Float ---")
print(float("123.45"))
print(type(float("123.45")))

print("\n--- Converting to String ---")
print(str(17))
print(type(str(17)))
print(str(123.45))
print(type(str(123.45)))
````

# Chapter 2.6: Statements and Expressions

Understanding the difference between a **statement** and an **expression** is vital for writing and debugging Python code.

### 1. Statements
A **statement** is a unit of code that the Python interpreter can execute. It represents an action or a command.
* So far, we have focused on the **assignment statement** (e.g., `x = 5`).
* Other statements you will learn later include `if`, `for`, `while`, and `import`.
* **Key Fact:** Statements do not return a value; they simply perform a task.

### 2. Expressions
An **expression** is a combination of values, variables, operators, and function calls.
* **Key Fact:** Expressions must be **evaluated** to produce a single value.
* Because expressions result in a value, they can appear on the right-hand side of an assignment statement.

````python
# Activity 2.6.2: Evaluating Expressions
# Python evaluates these and then the print function displays the result

print(1 + 1)           # Arithmetic expression
print(len("hello"))    # Function call expression (len returns the length)

# A variable by itself is also an expression
message = "Python"
print(message)         # Evaluating the variable 'message'
````

### 3. Combining Statements and Expressions
We often use expressions to define the value being assigned in a statement.

* **Statement:** `x = len("hello")`
* **Expression being evaluated:** `len("hello")`
* **Resulting value:** `5`

In the Python Shell (REPL), if you type an expression, it shows the result immediately. If you type a statement (like `y = 3.14`), the shell performs the action but shows no output until you ask it to evaluate `y`.

````python
# Activity 2.6.3: Assignment using expressions
y = 3.14             # Statement
x = len("hello")     # Statement containing an expression

print(x)             # Evaluating x
print(y)             # Evaluating y
````

# Chapter 2.7: Operators and Operands

**Operators** are special tokens that represent computations like addition, multiplication, and division. The values the operator works on are called **operands**.

### Standard Mathematical Operators
Python uses standard symbols for most math:
* `+` : Addition
* `-` : Subtraction
* `*` : Multiplication
* `**` : Exponentiation (Power)
* `/` : Division (Always results in a float/decimal)



### The Two Flavors of Division
1. **Floating-point Division (`/`)**: Returns the full decimal result (e.g., `7 / 4` is `1.75`).
2. **Integer Division (`//`)**: Truncates the result **down** to the next smallest integer (e.g., `7 // 4` is `1`).

### The Modulus Operator (`%`)
The **modulus** (or remainder) operator yields the remainder after integer division.
* Example: `7 % 3` is `1`, because 3 goes into 7 twice with 1 left over.

````python
# Activity 2.7.1: Basic Operators
print("Addition:", 2 + 3)
print("Subtraction:", 2 - 3)
print("Multiplication:", 2 * 3)
print("Power (2 to the 3rd):", 2 ** 3)
print("Power (3 squared):", 3 ** 2)

# Activity 2.7.2 & 2.7.3: Division Flavors
minutes = 645
print("\nStandard Division (/) results in float:", minutes / 60)
print("Integer Division (//) results in int:", minutes // 60)

# Note on negative integer division (truncates toward the left on number line)
print("\nInteger division with negatives:")
print("6 // 4  =", 6 // 4)
print("-6 // 4 =", -6 // 4)
````

### Use Cases for Modulus (%)
The `%` operator is more useful than it first appears:
* **Check Divisibility**: If `x % y` is `0`, then `x` is divisible by `y`.
* **Extract Digits**: `x % 10` gives you the last digit of a number.
* **Time Conversion**: Use it to find remaining seconds after calculating whole hours.

````python
# Activity 2.7.4 & 2.7.5: Modulus and Time Conversion
quotient = 7 // 3
remainder = 7 % 3
print("7 divided by 3 is", quotient, "with a remainder of", remainder)

# Converting 7684 seconds into Hours, Minutes, Seconds
total_secs = 7684
hours = total_secs // 3600
secs_still_remaining = total_secs % 3600
minutes =  secs_still_remaining // 60
secs_finally_remaining = secs_still_remaining  % 60

print("\n7684 seconds converts to:")
print(hours, "hours,", minutes, "minutes, and", secs_finally_remaining, "seconds")
````

# Chapter 2.8: Input

To make programs more general and useful, we often need to get information from the person using the program. Python provides a built-in function called `input()` for this purpose.

### How the `input()` Function Works:
1.  **The Prompt:** You can provide a "prompt string" inside the parentheses to tell the user what to type.
2.  **The Pause:** The program pauses and waits for the user to type something and press **Enter**.
3.  **The Return Value:** The function "returns" whatever the user typed as a **string**.



### The String Trap
It is critical to remember that **`input()` always returns a string (`str`)**, even if the user types a number like `17`. If you want to use that input for math, you must convert it using `int()` or `float()`.

````python
# Activity 2.8.2: Basic Input
n = input("Please enter your name: ")
print("Hello", n)

# Checking the type of an input
age = input("Please enter your age: ")
print("Your age is", age, "and its type is", type(age))
````

### Converting Input for Calculations
To use user input in our previous time conversion program, we have to wrap the input in a type converter. This allows the program to perform integer division and modulus operations.

````python
# Activity 2.8.3: General Time Converter
str_seconds = input("Please enter the number of seconds you wish to convert: ")

# Convert the string to an integer so we can do math
total_secs = int(str_seconds)

hours = total_secs // 3600
secs_still_remaining = total_secs % 3600
minutes =  secs_still_remaining // 60
secs_finally_remaining = secs_still_remaining  % 60

print("Hrs=", hours, "mins=", minutes, "secs=", secs_finally_remaining)
````

# Chapter 2.9: Order of Operations

When multiple operators appear in an expression, Python follows a specific order to decide which part to calculate first. This is known as **Rules of Precedence**. Python follows the same rules used in mathematics (often remembered by the acronym **PEMDAS**).

### The Order of Precedence (High to Low):

1.  **Parentheses `()`**: Highest precedence. Use these to force an order or make code readable.
2.  **Exponentiation `**`**: Next highest.
3.  **Multiplication `*`, Division `/`, Integer Division `//`, and Modulus `%`**: These all have the same level of precedence.
4.  **Addition `+` and Subtraction `-`**: Lowest mathematical precedence.



### Left-to-Right Associativity
When operators have the **same** precedence (like `+` and `-`), Python evaluates them from **left to right**.
* Example: `6 - 3 + 2` is evaluated as `(6 - 3) + 2 = 5`.

### The Exponentiation Exception
The exponentiation operator `**` is **right-associative**, meaning it evaluates from **right to left**.
* Example: `2 ** 3 ** 2` is evaluated as `2 ** (3 ** 2) = 2 ** 9 = 512`.

````python
# Activity 2.9.2: Order of Operations Examples

# Parentheses force the order
print("Parentheses:", 2 * (3 - 1))

# Exponentiation vs Multiplication
print("Exponentiation first:", 3 * 1 ** 3)

# Right-to-Left Exponentiation
print("Right-associative:", 2 ** 3 ** 2)   # 2 ** 9
print("Manual Parentheses:", (2 ** 3) ** 2) # 8 ** 2
````

# Chapter 2.10: Reassignment

In Python, it is perfectly legal to make more than one assignment to the same variable. This is called **reassignment**.

### How Reassignment Works
When you reassign a variable, it stops referring to its old value and starts referring to a new one.
* It is **not** like a permanent mathematical equation.
* In math, if $a = b$, it is always true. In Python, `a = b` means `a` currently refers to whatever `b` refers to, but that can change later.



### Developing a Mental Model
When Python sees an assignment statement like `a = 5 + 2`, it follows two steps:
1. **Evaluate the Right Side:** It calculates the result of the expression (e.g., `7`).
2. **Assign the Name:** It attaches the name on the left to that resulting value.

````python
# Activity 2.10.2: Simple Reassignment
bruce = 5
print("First value of bruce:", bruce)

bruce = 7
print("Second value of bruce:", bruce)
````

### Tracking Variable Changes
If you assign one variable to another (e.g., `b = a`), they both point to the same value. However, if you later change `a`, `b` does **not** automatically change with it. This is a common point of confusion for beginners!

````python
# Activity 2.10.3: Changing one variable but not the other
a = 5
b = a    # b now refers to the same 5 that a refers to
print("Initially equal:", a, b)

a = 3    # a is reassigned to 3, but b still refers to 5
print("After reassigning a:", a, b)
````

# Chapter 2.11: Updating Variables

One of the most common forms of reassignment is an **update**, where the new value of the variable depends on its old value.

### The Logic of `x = x + 1`
At first glance, this looks like a mathematical impossibility. However, remember that Python performs assignment in two distinct steps:
1.  **Evaluate the Right-Hand Side:** Python looks at the current value of `x` and adds 1 to it.
2.  **Update the Variable:** Python then takes that new result and assigns the name `x` to it.



### Initialization
You cannot update a variable that does not exist yet. You must **initialize** it first (give it a starting value) before you can use it on the right-hand side of an assignment.

### Vocabulary:
* **Increment:** Increasing the value of a variable (usually by 1).
* **Decrement:** Decreasing the value of a variable (usually by 1).
* **Bumping:** A common programmer term for incrementing by 1.

````python
# Activity 2.11.3: Incrementing a variable
x = 6        # Initialization
print("Initial x:", x)

x = x + 1    # Update/Increment
print("Updated x:", x)

# This would cause a NameError because 'y' isn't initialized:
# y = y + 1
````

# Chapter 2.12: Glossary

This glossary summarizes the key terms and concepts introduced in Chapter 2. Use this as a reference guide for your Python studies.

### Core Programming Concepts
* **Value:** A number, string, or other data that can be stored in a variable or computed in an expression.
* **Object:** Also known as a data value; the fundamental things programs manipulate.
* **Variable:** A name that refers to a value.
* **Reference Diagram:** A picture showing a variable with an arrow pointing to the value (object) it refers to. Also called a **State Snapshot**.



### Data Types
* **Data Type (Class):** A set of values. The type determines how a value can be used.
* **int:** A data type for whole numbers (positive or negative).
* **float:** A data type for decimal numbers (floating-point).
* **str:** A data type for a sequence of characters (strings).
* **Type Conversion Function:** Functions like `int()`, `float()`, and `str()` that change a value from one type to another.

### Statements and Expressions
* **Statement:** An instruction the Python interpreter can execute (e.g., assignment).
* **Expression:** A combination of operators and operands that represents a single result value.
* **Evaluate:** To simplify an expression down to a single value.
* **Assignment Statement:** A statement that assigns a value to a name.
* **Assignment Token (`=`):** The symbol used to link a name to a value.
* **Initialization:** Giving a variable its very first value.

### Operators and Math
* **Operator:** Special symbols for computations (like `+`, `-`, `*`, `**`).
* **Operand:** The values that an operator works on.
* **Integer Division (`//`)**: An operation that discards the remainder and yields a whole number.
* **Modulus Operator (`%`)**: Also called the remainder operator; gives the remainder of integer division.
* **Rules of Precedence:** The order in which Python evaluates operators (e.g., PEMDAS).
* **Increment / Decrement:** Increasing or decreasing a variable's value by 1.

### Language Rules
* **Keyword:** Reserved words (like `if`, `while`, `class`) that cannot be used as variable names.
* **Variable Name:** A sequence of letters, digits, and underscores starting with a letter.
* **Comment:** Information in a program meant for humans; it has no effect on execution.
* **Prompt String:** The text displayed by the `input()` function to guide the user.

````python
# Quick code review of the Glossary concepts

# 1. Initialization and Assignment
score = 0

# 2. Expression evaluation and Reassignment
# 'score + 10' is the expression; the whole line is a statement
score = score + 10

# 3. Type Conversion and Input
# prompt string: "Enter bonus: "
user_input = "5" # Simulating input
bonus = int(user_input) # Type conversion

# 4. Final Calculation
total = score + bonus
print("Final Score:", total)
````
