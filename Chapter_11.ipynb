# 11.1. Working with Data Files

Up to this point, we have either hard-coded data into our scripts or asked the user for input. However, real-world data typically lives in **files** (text documents, images, music, or spreadsheets). In this chapter, we focus on **text files**—files containing sequences of characters that Python can read and manipulate.

### The File Object Lifecycle
To use a file in Python, you must follow a specific process:
1.  **Open**: Create a connection between your Python program and the file on the disk.
2.  **Process**: Read data from the file or write data to it.
3.  **Close**: Disconnect the file. This is crucial because it frees up system resources and ensures all data is physically saved to the disk.



### Opening a File
We use the `open()` function, which requires two main pieces of information:
* **Filename**: The name (and path) of the file you want to access.
* **Mode**: A single character string telling Python what you intend to do.
    * `'r'`: **Read mode** (Default). Use this to look at the contents without changing them.
    * `'w'`: **Write mode**. Use this to create a new file or overwrite an existing one.



### File Methods Summary
| Method | Syntax | Explanation |
| :--- | :--- | :--- |
| **open** | `open(filename, 'r')` | Returns a **file object** reference for reading. |
| **open** | `open(filename, 'w')` | Returns a **file object** reference for writing. |
| **close** | `file_var.close()` | Finalizes the file use and releases it from memory. |

```python
# Demonstrating Opening and Closing Files

# Note: In a real environment like Colab, you must have the file
# existing in the file sidebar for 'r' mode to work.

# 1. Writing to a file
# Opening in 'w' mode creates the file if it doesn't exist.
my_file = open("test_data.txt", "w")
my_file.write("Hello Data!\nThis is our first file.")
my_file.close() # Always close after writing!
print("File created and closed successfully.")

# 2. Reading from a file
try:
    # Opening in 'r' mode (default)
    reader = open("test_data.txt", "r")
    content = reader.read()
    print("\nFile Contents:")
    print(content)
    reader.close() # Always close after reading!
except FileNotFoundError:
    print("The file was not found.")

# 3. Modern Pythonic Way (The Context Manager)
# This automatically closes the file for you even if an error occurs.
with open("test_data.txt", "r") as f:
    data = f.read()
    print("\nRead using 'with' statement:")
    print(data)
# No .close() needed here!
```

## Key Takeaways

* **File Objects**: When you `open()` a file, Python creates a "file object" (a reference). You call methods like `.read()` or `.write()` on this object.
* **The 'w' Danger**: Opening a file in `'w'` mode immediately **erases** everything in the existing file. If you want to add to a file instead of overwriting, you would use `'a'` (append mode).
* **Closing is Mandatory**: If you don't close a file, you might experience "data loss" where the information stays in a temporary buffer and never actually makes it to the hard drive.

> **Pro Tip**: Use the `with` statement (shown in the code cell above). It is the safest way to handle files because it guarantees the file will be closed automatically when the code block finishes.

# 11.2. Finding a File on your Disk

To open a file, Python needs to know its exact location on your computer. This location is called a **path**. You can think of a filename (e.g., `hello.txt`) as a "first name" and the path as the "full name."

### Absolute vs. Relative Paths

There are two ways to tell Python where a file is:

1.  **Absolute Path**: The complete "mailing address" starting from the very root of the hard drive (e.g., `C:/Users/name/file.txt` or `/Users/name/file.txt`). These are rarely used in shared programs because they only work on *your* specific computer.
2.  **Relative Path**: A path that starts from the folder where your Python script is currently running. This is the preferred method for most programming tasks.



### Navigating the Hierarchy
The file system is a tree-like structure of folders (directories). To navigate this tree in a relative path, we use specific symbols:
* **`folder_name/`**: Move **down** into a sub-folder.
* **`..`**: Move **up** one level to the "parent" folder.
* **`.`**: Refers to the "current" folder.



### The Path Separator "/"
While Windows traditionally uses backslashes (`\`) and Mac/Linux use forward slashes (`/`), Python is smart enough to handle the forward slash `/` on **all systems**. Using `/` is safer because `\` is the "escape character" in Python (like in `\n`), which often causes bugs in file paths.

```python
# Working with paths and directories
import os

# 1. Get the current working directory
# This tells you where Python is 'looking' right now
current_dir = os.getcwd()
print(f"Your Python script is running in: {current_dir}")

# 2. Example of path construction
# If your script is in /Project/
# and your data is in /Project/Data/files/
# A relative path would be:
relative_path = "Data/files/mydata.txt"

# 3. Using the '..' to go up
# If you need to reach a folder 'otherFiles' that sits next to 'Project'
# Path: ../otherFiles/info.txt
path_up = "../otherFiles/info.txt"

# 4. Checking if a file exists before opening
filename = "test_data.txt"
if os.path.exists(filename):
    print(f"Success! '{filename}' was found in the current directory.")
else:
    print(f"Warning: '{filename}' not found.")
```

## Path Rules of Thumb

* **Same Folder**: If the file and the `.py` script are in the same directory, just use the name: `open('data.txt', 'r')`.
* **Sub-folder**: Use the folder name: `open('logs/today.txt', 'r')`.
* **Parent Folder**: Use the double dots: `open('../config.txt', 'r')`.

### Summary of Path Components
| Component | Meaning | Example |
| :--- | :--- | :--- |
| **`filename.txt`** | Look in the current folder | `open('data.txt')` |
| **`folder/`** | Look inside a sub-folder | `open('images/cat.jpg')` |
| **`..`** | Go up to the parent folder | `open('../settings.py')` |
| **`/`** | The root of the drive (Absolute) | `open('/Users/home/file.txt')` |

# 11.3. Reading a File

Once a file is opened, we can use the **file object** to access the data stored inside. For example, if we have a dataset like `ccdata.txt` (Climate Change Data), we can interact with it using a reference variable.

### Understanding the Data Format
In the file `ccdata.txt`, each line represents a specific year and its associated metrics:
1. **Year**
2. **Global Average Temperature**
3. **Global Emission of CO2**

The data is organized in columns, often separated by spaces or tabs. This is a common format for scientific data.

### The File Reference
When you call `fileref = open("ccdata.txt", "r")`, `fileref` doesn't contain the actual text of the file. Instead, it is a **pointer** or a "handle" that allows you to request data from the file.



### Opening and Closing
It is vital to close the file after use. Once `fileref.close()` is called, the connection is severed. If you try to read from `fileref` afterward, Python will raise an `I/O Error` because the "book" has been closed and put back on the shelf.

```python
# Preparing the data environment
# Since we are in a notebook, we first create the file for demonstration
data_content = """1850 -0.37 2.24E-7
1860 -0.34 3.94E-7
1870 -0.28 6.6E-7
1880 -0.24 1.1
1890 -0.42 1.72
1900 -0.2 2.38"""

with open("ccdata.txt", "w") as f:
    f.write(data_content)

# --- Standard File Reading Pattern ---

# 1. Open the file for reading ('r')
fileref = open("ccdata.txt", "r")

# 2. At this point, we have a file object, but we haven't 'read' it yet.
print(f"File object created: {fileref}")

# 3. Close the file
fileref.close()

# 4. Attempting to use it after closing will cause an error
try:
    print(fileref.read())
except ValueError as e:
    print(f"\nCaught Error: {e}")
```

## File Reading Methods

While `open()` prepares the file, there are several ways to actually pull the data into your program. We will explore these in the next sections:

1.  **`.read()`**: Reads the entire file as one giant string.
2.  **`.readline()`**: Reads just the next line of the file.
3.  **`.readlines()`**: Reads all lines and puts them into a list of strings.
4.  **`for line in fileref`**: The most memory-efficient way to process a file line-by-line.



### Summary of the Climate Data
Notice the values in the third column: `2.24E-7`. This is **Scientific Notation**. When we eventually read this into Python, we will need to convert these strings into `float` types so we can perform math on them.

# 11.4. Iterating over Lines in a File

A text file is essentially a **sequence of lines**. Because of this, Python allows us to use a `for` loop to process a file one line at a time. This is the most efficient way to handle large files because Python only keeps one line in memory at any given moment.

### The Newline Character (`\n`)
Each line in a file ends with a special hidden character called the **newline** character, represented as `\n`.
* When you press **Enter** while typing, you are inserting a `\n`.
* **Crucial Difference**: Unlike the `input()` function (which strips the newline), reading from a file **includes** the `\n` at the end of the string.



### The Processing Pattern
To extract data from a file like `ccdata.txt`, we follow this pattern:
1.  **Iterate**: Use `for line in file_object:`
2.  **Split**: Use `line.split()` to turn the string into a list of "words" or data fields.
3.  **Process**: Access the data using indices (e.g., `values[0]` for the year).



### Formatting with `split()`
One of the best features of `.split()` is that it automatically ignores whitespace, including the trailing `\n`. This makes it very easy to clean up data from scientific files where columns might be separated by multiple spaces.

```python
# Creating the data file for the demonstration
data = """1850 -0.37 2.24E-7
1860 -0.34 3.94E-7
1870 -0.28 6.6E-7
1880 -0.24 1.1
1890 -0.42 1.72
1900 -0.2 2.38"""

with open("ccdata.txt", "w") as f:
    f.write(data)

# --- Iterating over the file ---

# Open the file for reading
ccfile = open("ccdata.txt", "r")

# The loop variable 'aline' will get one line per iteration
for aline in ccfile:
    # Split the line into a list of strings
    # Values will be: ['1850', '-0.37', '2.24E-7']
    values = aline.split()

    # Process the values (index 0=Year, 1=Temp, 2=CO2)
    year = values[0]
    temp = values[1]
    co2 = values[2]

    print(f"In {year}, average temp was {temp}°C and CO2 emissions were {co2} gigatons.")

# Remember to close the file!
ccfile.close()
```

## Check Your Understanding

**1. What happens if you print a line read from a file directly?**
* **Answer**: You will often see an "extra" blank line. This is because the line from the file ends with a `\n`, and the `print()` function adds its own `\n` by default.

**2. Why is `aline.split()` better than manually cutting the string?**
* **Answer**: It handles any amount of whitespace. Whether there is 1 space or 20 spaces between the columns in `ccdata.txt`, `split()` will reliably return the same list of data.

**3. If the last line of a file doesn't have a newline character, will the `for` loop still find it?**
* **Answer**: Yes. Python is designed to treat the final segment of text as a line, even if it lacks the terminating `\n`.

# 11.5. Alternative File Reading Methods

While the `for` loop is often the most convenient way to read a file, Python provides several other methods depending on how you need to access the data.

### The Read Position (The Marker)
Every open file has an internal **marker** (or cursor) that keeps track of where the next read operation will begin. When you read from a file, this marker moves forward. To start reading from the beginning again, you must either close and reopen the file or use a special method to "seek" back to the start.



### Reading Methods Summary

| Method | Returns | Best Used When... |
| :--- | :--- | :--- |
| **`.read()`** | A single **string** | You need the entire file content at once (and it fits in memory). |
| **`.readline()`** | One **string** (the next line) | You want to process the file line-by-line manually. |
| **`.readlines()`** | A **list of strings** | You need to access lines by index (e.g., `lines[5]`). |

### The `while` Loop Pattern
In many programming languages, a `while` loop is the standard way to read files. This uses a **priming read**—you read the first line before the loop starts to give the `while` condition something to check. Inside the loop, you must perform another read to move the marker to the next line.

```python
# Setup: Ensuring the data file exists
data = """1850 -0.37 2.24E-7
1860 -0.34 3.94E-7
1870 -0.28 6.6E-7"""
with open("ccdata.txt", "w") as f:
    f.write(data)

# --- 1. Using readline() with a while loop ---
print("--- While Loop Method ---")
infile = open("ccdata.txt", "r")
line = infile.readline()  # Priming read
while line:
    values = line.split()
    print(f"Year {values[0]}: {values[1]} °C")
    line = infile.readline() # Change of state: read next line
infile.close()

# --- 2. Using readlines() to get a list ---
print("\n--- Readlines Method ---")
infile = open("ccdata.txt", "r")
all_lines = infile.readlines()
infile.close()

print(f"Total lines: {len(all_lines)}")
print(f"The second line is: {all_lines[1].strip()}")

# --- 3. Using read() for a single string ---
print("\n--- Read Method ---")
infile = open("ccdata.txt", "r")
content = infile.read()
infile.close()
print(f"Total characters in file: {len(content)}")
```

## Check Your Understanding

**1. Why does `while line:` stop when the end of the file is reached?**
* **Answer**: At the end of the file, `.readline()` returns an **empty string** (`""`). In Python, an empty string is considered `False`, which terminates the loop. Note that a blank line in a file is *not* empty; it contains `\n`, which is `True`.

**2. What is a "priming read"?**
* **Answer**: It is the first read operation performed before entering a loop. It ensures the loop variable is initialized so the condition can be tested the first time.

**3. If you call `.read()` and then immediately call `.readlines()`, what will happen?**
* **Answer**: The `.readlines()` call will return an empty list. This is because `.read()` moves the file marker all the way to the end of the file; there is nothing left to read unless you reopen the file or reset the marker.

# 11.6. Writing Text Files

The power of data processing comes from the ability to read an existing file, transform the data, and save the results into a **new file**. This allows you to create summaries, filtered datasets, or reports that can be used later by other programs.

### Opening for Writing
To create a new file, we use the `open()` function with the `'w'` flag:
`outfile = open("new_file.txt", "w")`

**⚠️ Important Warnings:**
* **Overwrite**: If a file with that name already exists, opening it in `'w'` mode will immediately **erase** all its current contents.
* **Creation**: If the file does not exist, Python will create a brand-new empty file for you.

### The `write` Method
The `.write()` method takes exactly one argument: a **string**.
Unlike the `print()` function, `.write()` **does not** automatically add a newline character to the end. As the programmer, you must manually concatenate `\n` if you want the next piece of data to appear on a new line.



### Transformation Workflow
A common "Data Cleaning" pattern looks like this:
1. **Open** the source file (`'r'`) and a destination file (`'w'`).
2. **Loop** through the source lines.
3. **Parse** and **Filter** the data (using `.split()`).
4. **Write** the selected fields to the destination file.
5. **Close** both files to ensure data is saved.

```python
# Setup: Create dummy climate data
with open("ccdata.txt", "w") as f:
    f.write("1850 -0.37 2.24E-7\n1860 -0.34 3.94E-7\n1870 -0.28 6.6E-7")

# --- DATA TRANSFORMATION TASK ---
# Goal: Create a new file with only Year and Emissions, separated by a tab.

infile = open("ccdata.txt", "r")
outfile = open("emissiondata.txt", "w")

# Add a header to our new file
outfile.write("Year\tEmission\n")

# Process the file line by line
aline = infile.readline()
while aline:
    items = aline.split()

    # Extract Year (index 0) and Emission (index 2)
    # items[1] (Temperature) is skipped!
    dataline = items[0] + '\t' + items[2]

    # Write to file - manually adding the newline!
    outfile.write(dataline + '\n')

    # Move to next line
    aline = infile.readline()

# CRITICAL: Always close both!
infile.close()
outfile.close()

# Verify the result
with open("emissiondata.txt", "r") as check:
    print("Contents of emissiondata.txt:")
    print(check.read())
```

## Check Your Understanding

**1. What happens if you forget to add `+ '\n'` in your `outfile.write()` call?**
* **Answer**: All your data will be smashed together on one single, very long line. This makes the file very difficult for humans (and other programs) to read.

**2. Why do we need two file variables (`infile` and `outfile`)?**
* **Answer**: Each variable tracks its own "marker" or position. `infile` tracks where we are reading from, and `outfile` tracks where we are writing to.

**3. What is the difference between `\t` and `\n`?**
* **Answer**: `\n` is a **newline** (moves to the next row), while `\t` is a **tab** (pushes text to the next horizontal alignment stop, creating columns).

# 11.7. With Statements (Context Managers)

In professional Python development, the most common way to handle files is not by calling `.close()` manually, but by using the **`with` statement**. This statement uses what is known as a **Context Manager** to handle the "setup" and "teardown" logic of a task automatically.

### Why Use `with`?
When you use a `with` statement, Python guarantees that the file will be closed the moment the code block ends. This is important because:
* **Error Safety**: If your code crashes inside the block, the file is still safely closed.
* **No Memory Leaks**: It prevents "hanging" file handles that can slow down or crash your computer if you open thousands of files without closing them.
* **Cleaner Code**: You don't have to remember to type `.close()` at the end of every function.



### The Syntax
The `with` statement follows this structure:
`with open(filename, mode) as <variable_name>:`
    # Everything indented here is "inside" the context
    # Use <variable_name> to read or write

# Once you un-indent, the file is automatically closed!

```python
# Demonstrating the 'with' statement

# 1. Create a sample file
with open('mydata.txt', 'w') as writer:
    writer.write("1 2 3\n4 5 6")

# 2. Reading with 'with'
print("--- Reading inside the 'with' block ---")
with open('mydata.txt', 'r') as md:
    print(f"File status inside block: Is it closed? {md.closed}")
    for line in md:
        print(f"Line content: {line.strip()}")

# 3. Checking the status outside the block
# Notice we don't call md.close()!
print(f"\nFile status outside block: Is it closed? {md.closed}")

# 4. Processing both Input and Output simultaneously
# You can even open two files at once in one 'with' statement!
try:
    with open('mydata.txt', 'r') as infile, open('copy.txt', 'w') as outfile:
        for line in infile:
            outfile.write("Copy: " + line)
    print("\nSimultaneous read/write successful.")
except FileNotFoundError:
    print("\nFile not found.")
```

## Check Your Understanding

**1. What happens to the file if the program crashes while inside a `with` block?**
* **Answer**: The Context Manager ensures the file is closed safely before the program exits, preventing data corruption.

**2. Can you use the file variable (like `md`) after the `with` block has ended?**
* **Answer**: You can still refer to the variable, but any attempt to **read** or **write** to it (e.g., `md.read()`) will raise a `ValueError` because the file is already closed.

**3. Is it "wrong" to use `.close()`?**
* **Answer**: No, it isn't wrong, but using `with` is considered "Pythonic" and best practice. It makes your code more robust and readable.

# 11.8. Fetching Something From The Web

Python's ability to handle files isn't limited to the data already sitting on your hard drive. You can also open "files" located on the internet using a URL (Uniform Resource Locator).

### How Web Fetching Works
When you visit a website, your browser performs a "GET" request to a server. In Python, we can do the same thing using the `urllib.request` module.
* Instead of a filename on your disk, you provide a **URL**.
* Instead of `open()`, we use `urllib.request.urlopen()`.
* This creates a **socket** (a network connection) that acts very much like a file object.

### The Binary Difference
When you read data from the web using `.read()`, Python often returns it in **bytes** (binary data) rather than a standard string. You will see a `b` prefixing the data (e.g., `b'<html>'`). To turn this into a readable string, you usually need to `.decode()` it.



### Key Requirements
1. **Connectivity**: You must have an active internet connection.
2. **Resource Existence**: The URL must be valid and the resource must still exist.
3. **Permissions**: Some websites block "bots" or scripts from reading their content to prevent scraping.

```python
# Demonstrating Web Data Retrieval
import urllib.request

def retrieve_page(url):
    """ Retrieve the contents of a web page and return as a string. """
    try:
        # 1. Open a connection to the URL (like opening a file)
        my_socket = urllib.request.urlopen(url)

        # 2. Read the data (returns bytes)
        raw_data = my_socket.read()

        # 3. Decode bytes into a UTF-8 string
        decoded_text = raw_data.decode("utf-8")

        # 4. Close the connection
        my_socket.close()

        return decoded_text
    except Exception as e:
        return f"Error retrieving page: {e}"

# Example usage with a sample Runestone URL
target_url = "https://runestone.academy/runestone/static/thinkcspy/_static/testfile.txt"
# Note: If the specific URL in the book is down, we use a generic placeholder
the_text = retrieve_page(target_url)

print("First 200 characters of retrieved content:")
print("-" * 40)
print(the_text[:200])
print("-" * 40)
```

## Check Your Understanding

**1. What is the difference between `open()` and `urllib.request.urlopen()`?**
* **Answer**: `open()` looks for a file on your local storage (hard drive). `urlopen()` looks for a resource on a remote server over the internet.

**2. Why do we see a `b` before the text when we print web data?**
* **Answer**: The `b` stands for **bytes**. Data travels across the internet as a series of 0s and 1s. You must use `.decode()` to translate those bytes back into human-readable characters (strings).

**3. Can you use a `for` loop to iterate over a web page?**
* **Answer**: Yes! Just like a local file, a web socket is iterable. You can use `for line in my_socket:` to process a web page line by line.

# 11.9. Glossary

This glossary summarizes the core concepts of file handling and navigation. Mastering these terms is key to managing data outside of your Python script's memory.

### File Operations
| Term | Definition |
| :--- | :--- |
| **open** | The initial step required to create a connection between Python and a file on disk. |
| **close** | The final step that severs the connection, saves data, and frees system resources. |
| **read** | A method that pulls the entire file's content into a single string. |
| **readline** | A method that reads the file one line at a time (up to the next `\n`). |
| **readlines** | A method that reads the entire file and organizes it into a **list of strings**. |
| **write** | A method that adds a string to a file opened in `'w'` or `'a'` mode. |



### File Navigation (Paths)
* **Absolute File Path**: The full address of a file starting from the **root** of the computer. It always starts with a `/` (on Unix/Mac) or a drive letter like `C:/` (on Windows).
* **Relative File Path**: The address of a file starting from the **Current Working Directory** (where your script is saved). It is more portable than an absolute path.



### Core Concept: The Newline Character
A invisible character (`\n`) that marks the end of a line. `readline` and `readlines` include this character at the end of their strings, while `split()` ignores it.

```python
# Glossary Demonstration: Paths and Methods

import os

# 1. Understanding Paths
# Relative Path: Just the name (if in same folder)
rel_path = "my_data.txt"

# Absolute Path: Derived from current location
abs_path = os.path.abspath(rel_path)
print(f"Relative Path: {rel_path}")
print(f"Absolute Path: {abs_path}")

# 2. Comparing Read Methods
with open(rel_path, "w") as f:
    f.write("Line 1\nLine 2\nLine 3")

# .read() -> returns one string
with open(rel_path, "r") as f:
    print(f"\n.read() output: {repr(f.read())}")

# .readlines() -> returns a list
with open(rel_path, "r") as f:
    print(f".readlines() output: {f.readlines()}")

# .readline() -> returns first line string
with open(rel_path, "r") as f:
    print(f".readline() output: {repr(f.readline())}")
```
