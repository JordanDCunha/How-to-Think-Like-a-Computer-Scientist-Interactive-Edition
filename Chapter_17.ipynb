# 17.1. Object-Oriented Programming (OOP)

Python is an **object-oriented programming language**. While we have spent most of our time using **Procedural Programming**, it is time to shift our perspective toward **OOP**.

### Procedural vs. Object-Oriented
* **Procedural Programming**: The focus is on writing **functions** (procedures) that operate on data. Data and logic are often kept separate.
* **Object-Oriented Programming**: The focus is on creating **objects** that bundle both **data** (attributes) and **functionality** (methods) together into a single unit.



### Why use OOP?
OOP was developed in the 1960s but became the industry standard in the 1980s to solve specific problems:
1.  **Complexity**: As software grew, keeping track of hundreds of independent functions became impossible.
2.  **Modularity**: Objects act like "black boxes." You can change how an object works internally without breaking the rest of your program.
3.  **Real-world Modeling**: OOP allows us to model programs after real-world concepts. For example, a `Car` object has data (color, fuel level) and behaviors (drive, honk).



### Key Terminology
* **Class**: The blueprint or template for an object (e.g., the concept of a "Dog").
* **Object (Instance)**: A specific realization of a class (e.g., your specific dog "Fido").
* **Attributes**: The data stored inside an object (variables).
* **Methods**: The actions an object can perform (functions).

````python
# Procedural vs Object-Oriented Thinking

# --- 1. Procedural Way ---
# Data and functions are separate
dog_name = "Fido"
dog_breed = "Collie"

def bark(name):
    print(f"{name} says Woof!")

bark(dog_name)


# --- 2. Object-Oriented Way ---
# Data and functions are bundled together
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        print(f"{self.name} the {self.breed} says Woof!")

# Create an 'instance' of the Dog class
my_dog = Dog("Fido", "Collie")
my_dog.bark()
````

* **Answer**: It organizes code into manageable, reusable pieces. Instead of a "spaghetti" of functions, you have distinct objects that interact with each other in predictable ways.

**3. True or False: In Procedural programming, the focus is on the data itself rather than the functions.**

# 17.2. A Change of Perspective

Moving from procedural to object-oriented programming (OOP) requires a shift in how you view the "actors" in your code. It is a transition from focusing on **actions** to focusing on **entities**.

### The "Active Agent"
In procedural programming, the **function** is the active agent. You tell a function to do something and give it data to work on:
* *Procedural style:* `draw_square(my_turtle)` — "Hey function, take this turtle and make it draw."

In OOP, the **object** is the active agent. Objects have their own internal capabilities:
* *Object-Oriented style:* `my_turtle.forward(100)` — "Hey turtle, use your own internal ability to move forward."



### Real-World Modeling
The most significant advantage of OOP is that it mirrors our physical reality. We don't have a universal "heat" function in the middle of our kitchen that we shove a toaster into; instead, the toaster has its own internal "toast" method.

**Mental Chunking:**
By grouping data (state) and functions (behavior) together, we create "chunks" that are easier for our brains to manage.
* **A Microwave Object:**
    * **Data:** Timer, Power Level, Door Status.
    * **Methods:** `start()`, `stop()`, `set_timer()`.



### Why Politeness Matters (Versatility)
By making the object responsible for the action, we can write more flexible code. If you have different types of "Drawers" (a Turtle, a Pen, a Brush), you can simply tell each one to `.draw()`. You don't need a massive procedural function that has to figure out which type of object it was just handed.

````python
# Shifting Perspectives: Procedural vs. OOP

# 1. The Procedural Perspective (Action-oriented)
def calculate_area(radius):
    return 3.14159 * (radius ** 2)

circle_radius = 5
# We pass data into a standalone function
area = calculate_area(circle_radius)
print(f"Procedural Area: {area}")


# 2. The OOP Perspective (Object-oriented)
class Circle:
    def __init__(self, radius):
        self.radius = radius # The object "owns" its data

    def area(self): # The object "owns" the logic
        return 3.14159 * (self.radius ** 2)

# We create an agent (my_circle) and ask IT to perform the task
my_circle = Circle(5)
print(f"OOP Area: {my_circle.area()}")
````

## Key Takeaways

1.  **Responsibility**: OOP shifts the responsibility of performing a task from a global function to the specific object involved.
2.  **Encapsulation**: This is the fancy word for "bundling." We encapsulate the radius and the area logic inside the `Circle` object.
3.  **Intuition**: It is often easier to design a system by asking "What things are in this program?" rather than "What actions are happening?"

### Comparison Table
| Feature | Procedural | Object-Oriented |
| :--- | :--- | :--- |
| **Primary Focus** | Logic/Algorithms | Data/Entities |
| **Syntax** | `verb(noun)` | `noun.verb()` |
| **Organization** | Functions and variables | Classes and objects |
| **Real-world analogy** | A recipe book | A kitchen full of appliances |

# 17.3. Objects Revisited

In Python, the phrase **"Everything is an object"** is literally true. Whether you are working with a complex Turtle, a List of numbers, or even a simple integer like `5`, you are interacting with an object.

### The Two Pillars of an Object: State and Methods
An object is defined by two main characteristics:

1.  **State (Attributes)**: These are the things the object **knows** about itself.
    * *Example*: A Turtle knows its current $x$ and $y$ coordinates, its color (e.g., "blue"), and its heading (e.g., 90 degrees).
2.  **Methods (Behaviors)**: These are the things the object can **do**.
    * *Example*: A Turtle can move `forward()`, turn `left()`, or lift its `penup()`.



### Different Objects of the Same Type
While two objects might belong to the same category (like two different Turtle objects), they are unique because their **state** can be different.
* Turtle A might be at $(0, 0)$ and be Red.
* Turtle B might be at $(100, 50)$ and be Green.

They both have the same *ability* to move, but they exist as distinct individuals in the computer's memory.



### Built-in Objects
Even basic types in Python follow this rule:
* **Lists**: Their *state* is the items they hold; their *methods* include `.append()` and `.sort()`.
* **Strings**: Their *state* is the characters in the text; their *methods* include `.upper()` and `.find()`.

````python
# Demonstrating that everything is an object

# 1. An Integer is an object
# It has state (the value 5) and methods!
x = 5
print(f"Is x an instance of object? {isinstance(x, object)}")
# Integers even have methods like bit_length()
print(f"Bit length of the integer {x}: {x.bit_length()}")

# 2. A List is an object
# State: the numbers 1, 2, 3
my_list = [3, 1, 2]
# Method: .sort() changes the state of the object
my_list.sort()
print(f"Updated list state: {my_list}")

# 3. Custom objects (Turtles-style thinking)
class SimpleTurtle:
    def __init__(self, color):
        self.color = color  # State
        self.position = 0   # State

    def forward(self, distance): # Method
        self.position += distance
        print(f"The {self.color} turtle is now at position {self.position}")

t1 = SimpleTurtle("Red")
t2 = SimpleTurtle("Blue")

t1.forward(10)
t2.forward(5)
````

| **Object** | The specific instance | `tess` or `alex` |
| **State** | Data/Knowledge | `color`, `position`, `heading` |
| **Methods** | Actions/Functions | `forward()`, `right()` |

### Summary
Every value in Python is an object that encapsulates its own data (state) and the functions that can manipulate that data (methods). Shifting from seeing variables as "just data" to seeing them as "active objects" is the key to mastering Python.

# 17.4. User Defined Classes

While Python provides built-in classes like `str` and `Turtle`, the real power of programming comes from creating your own **User Defined Classes** to model specific problems, such as a mathematical `Point`.

### Defining the Class
A class definition acts as a **blueprint** or a factory. It defines what data the objects will hold and what actions they can perform.
* **Header**: Starts with the `class` keyword and the name (usually capitalized).
* **Docstring**: An optional string describing the class purpose.



### The Initializer: `__init__`
Every class should have a special method named `__init__`.
* **The Constructor**: This method is automatically called when you create a new instance.
* **The `self` Parameter**: This refers to the specific instance being created. It allows you to attach data (attributes) to that specific object.
* **Attributes**: Variables like `self.x` and `self.y` that store the object's state.



### Instantiation
When you call `Point()`, you are using a **constructor**. This triggers two steps:
1.  **Creation**: A new, empty object is made in memory.
2.  **Initialization**: The `__init__` method runs to set the "factory default" settings.

````python
# Defining and using our first custom class

class Point:
    """ Point class for representing and manipulating x,y coordinates. """

    def __init__(self):
        """ Create a new point at the origin (0,0) """
        # Attributes are attached to 'self'
        self.x = 0
        self.y = 0

# --- Instantiation ---

p = Point()         # Instantiate an object of type Point
q = Point()         # Instantiate a second, distinct Point object

# Each object has its own set of attributes
print(f"Point p is at: ({p.x}, {p.y})")
print(f"Point q is at: ({q.x}, {q.y})")

# They are identical in state, but different in identity
print(f"Is p the same object as q? {p is q}")
print(f"Are they the same type? {type(p) == type(q)}")

# We can modify an instance's state directly
p.x = 10
print(f"\nAfter update, p is at: ({p.x}, {p.y})")
print(f"q remains unchanged at: ({q.x}, {q.y})")
````

# 17.5. Improving our Constructor

Our initial version of the `Point` class was limited because every point started at `(0, 0)`. To make our class useful, we need the ability to specify the starting position when we create the object.

### Adding Parameters to `__init__`
Since `__init__` is a function, we can add parameters to it just like any other function. These parameters allow us to pass data into the "factory" so the object can be set up specifically.

1.  **The Parameters**: We add `initX` and `initY` to the method definition.
2.  **The Assignment**: Inside the method, we take those values and assign them to `self.x` and `self.y`.
3.  **The Call**: When we call the constructor `Point(7, 6)`, Python automatically passes `7` to `initX` and `6` to `initY`.



### Why use `initX` vs `self.x`?
It is important to distinguish between the **parameter** (the temporary value passed in) and the **attribute** (the variable that lives inside the object).
* `initX` is a local variable that exists only while the `__init__` function is running.
* `self.x` is a permanent part of the object's state.

````python
# Improving the Point class with custom initialization

class Point:
    """ Point class for representing and manipulating x,y coordinates. """

    def __init__(self, initX, initY):
        """ Create a new point at the given coordinates. """
        self.x = initX
        self.y = initY

# --- Instantiation with Arguments ---

# Now we can create points anywhere in 2D space!
p = Point(7, 6)
q = Point(3, 10)
origin = Point(0, 0)

print(f"Point p is at: ({p.x}, {p.y})")
print(f"Point q is at: ({q.x}, {q.y})")

# We can still perform math with these attributes
distance_x = p.x - q.x
print(f"The horizontal distance between p and q is: {abs(distance_x)}")
````

# 17.6. Adding Other Methods to our Class

The real "organizational power" of a class comes from grouping data with the **methods** that act upon that data. While a simple tuple like `(7, 6)` can store coordinates, it cannot calculate its own distance from the origin or move itself. A `Point` object can.

### What is a Method?
A method is a function that belongs to a specific class.
* **Invocation**: Methods are called using dot notation: `object.method()`.
* **The `self` Parameter**: Just like `__init__`, every method must have `self` as its first parameter. This allows the method to access the specific object's state (its `x` and `y` values).
* **Arguments**: When you call a method like `p.getX()`, you don't pass an argument for `self`. Python automatically passes the object `p` into that first slot.



### Why use Getters?
Methods like `getX` and `getY` are often called **accessor methods** or "getters." They provide a clean way to ask an object for its information. While you *can* access `p.x` directly, using a method allows you to change how that data is stored or calculated later without breaking the code that uses the class.

### Complex Logic: `distanceFromOrigin`
Methods can perform calculations using the object's attributes. To find the distance from $(0,0)$, we use the Pythagorean theorem:
$\sqrt{x^2 + y^2}$
In Python, this is written as:
`((self.x ** 2) + (self.y ** 2)) ** 0.5`

````python
# Extending the Point class with custom methods

class Point:
    """ Point class for representing and manipulating x,y coordinates. """

    def __init__(self, initX, initY):
        self.x = initX
        self.y = initY

    def getX(self):
        """ Return the x-coordinate of the point """
        return self.x

    def getY(self):
        """ Return the y-coordinate of the point """
        return self.y

    def distanceFromOrigin(self):
        """ Calculate the distance from (0,0) using Pythagorean theorem """
        return ((self.x ** 2) + (self.y ** 2)) ** 0.5

# --- Using the new methods ---

p = Point(7, 6)

print(f"X-coordinate: {p.getX()}")
print(f"Y-coordinate: {p.getY()}")
print(f"Distance from origin: {p.distanceFromOrigin():.2f}")

# Every instance has its own state and results
q = Point(3, 4)
print(f"Distance for q(3,4): {q.distanceFromOrigin()}") # Should be 5.0
````

# 17.7. Objects as Arguments and Parameters

Objects are data, and just like integers, strings, or lists, they can be passed as arguments into functions. When you pass an object to a function, the function receives a reference to that specific instance, allowing it to access the object's state and methods.

### Functions vs. Methods
It is important to distinguish between a **method** (which lives inside a class) and a **standalone function** (which lives outside).

1.  **Indentation**: Standalone functions are not indented inside the `class` block.
2.  **The `self` Parameter**: Standalone functions do not use `self`. Instead, they use descriptive parameter names for the objects they receive (like `point1` or `target_turtle`).
3.  **Invocation**: You call a function normally: `distance(p, q)`. You call a method with a dot: `p.distanceFromOrigin()`.



### How it Works
When the `distance` function is called:
* `point1` becomes an alias for the object `p`.
* `point2` becomes an alias for the object `q`.
* The function "asks" the objects for their coordinates using their internal `getX()` and `getY()` methods to perform the calculation.

````python
import math

class Point:
    """ Point class for representing and manipulating x,y coordinates. """

    def __init__(self, initX, initY):
        self.x = initX
        self.y = initY

    def getX(self):
        return self.x

    def getY(self):
        return self.y

# --- Standalone Function (Outside the Class) ---

def distance(point1, point2):
    """ Calculate the distance between two Point objects. """
    # We use the object's methods to get data
    xdiff = point2.getX() - point1.getX()
    ydiff = point2.getY() - point1.getY()

    # Apply the distance formula
    dist = math.sqrt(xdiff**2 + ydiff**2)
    return dist

# --- Main Program ---

p = Point(4, 3)
q = Point(0, 0)

# We pass the objects p and q as arguments
result = distance(p, q)

print(f"The distance between the points is: {result}")
````

## Check Your Understanding

**1. Is `distance` a method of the `Point` class?**
* **Answer**: No. It is a standalone function because it is not defined inside the `class Point:` block and does not take `self` as its first parameter.

**2. Could `distance` have been written as a method?**
* **Answer**: Yes! If it were a method, you would call it as `p.distanceTo(q)`. Inside the method, `self` would represent `p`, and you would pass `q` as a separate parameter.

**3. What happens if you pass something that isn't a Point (like an integer) into the `distance` function?**
* **Answer**: The program will crash with an `AttributeError`. Python will try to call `.getX()` on the integer, but integers do not have a `getX` method.

# 17.8. Converting an Object to a String

When you try to `print()` a custom object in Python, the default output isn't very helpful. It usually looks like `<__main__.Point object at 0x7f...>`, which tells you the class name and its memory address, but nothing about its **state** (the actual $x$ and $y$ values).

### The `__str__` Special Method
To fix this, Python allows you to **override** the default string behavior using a special method called `__str__`. Like `__init__`, it has double underscores on both sides.



* **Responsibility**: The `__str__` method must **return** a string.
* **Automatic Trigger**: This method is called automatically whenever you pass your object to `print()` or use the `str()` type converter.
* **Customization**: You, the programmer, decide exactly how the object should be formatted for a human reader.



### Why Override?
Overriding a method means providing a new implementation for a method that already exists in the background. By overriding `__str__`, you make debugging much easier because you can see the object's data clearly in the console.

````python
# Overriding the __str__ method for better output

class Point:
    """ Point class for representing and manipulating x,y coordinates. """

    def __init__(self, initX, initY):
        self.x = initX
        self.y = initY

    def getX(self):
        return self.x

    def getY(self):
        return self.y

    # This method is called by the print() function
    def __str__(self):
        # We must return a string!
        return f"Point({self.x}, {self.y})"

# --- Testing the output ---

p = Point(7, 6)
q = Point(-3, 10)

# Without __str__, this would print a cryptic memory address
# With __str__, it prints exactly what we defined above
print(f"Object p: {p}")
print(f"Object q: {q}")

# The str() function also uses our custom method
s = str(p)
print(f"The string version of p is: {s}")
````

## Check Your Understanding

**1. What is the return type of the `__str__` method?**
* **Answer**: It **must** return a string. If you try to return an integer or `None`, Python will raise a `TypeError`.

**2. What does it mean to "override" a method?**
* **Answer**: It means providing your own version of a special method to replace the default one provided by Python.

**3. If you have a class `Book` with attributes `title` and `author`, how might you write the `__str__` method?**
* **Answer**:
    ```python
    def __str__(self):
        return f"'{self.title}' by {self.author}"
    ```

# 17.9. Instances as Return Values

Methods are not limited to returning simple data types like integers or strings. A method can create a **brand new instance** of a class and return that entire object to the caller.

### Creating New Objects Inside Methods
In this example, we want to find the **midpoint** between two points.
1. The `halfway` method is called on one point (`self`).
2. It takes another point (`target`) as an argument.
3. It calculates the average $x$ and $y$ coordinates.
4. **Crucially**, it calls the `Point(mx, my)` constructor to build a third, new point and returns it.



### Disambiguation with Dot Notation
One of the best features of OOP is clarity. Inside the method:
* `self.x` clearly refers to the x-coordinate of the point the method was called on.
* `target.x` clearly refers to the x-coordinate of the point passed in as an argument.

There is no confusion about which data belongs to which entity.

### The Power of Returned Objects
Because the returned value is a `Point` object, you can immediately use any other `Point` methods on it, such as `getX()` or `distanceFromOrigin()`.

````python
# Returning a new Point instance from a method

class Point:
    def __init__(self, initX, initY):
        self.x = initX
        self.y = initY

    def __str__(self):
        return f"Point({self.x}, {self.y})"

    def getX(self):
        return self.x

    def getY(self):
        return self.y

    def halfway(self, target):
        """ Calculate the midpoint and return a NEW Point object """
        mx = (self.x + target.x) / 2
        my = (self.y + target.y) / 2

        # We create and return a brand new instance
        return Point(mx, my)

# --- Execution ---

p = Point(3, 4)
q = Point(5, 12)

# mid becomes a brand new Point object created by p
mid = p.halfway(q)

print(f"Point P: {p}")
print(f"Point Q: {q}")
print(f"Midpoint: {mid}")

# We can call methods on the returned object immediately
print(f"The midpoint's x-value is {mid.getX()}")
````

## Check Your Understanding

**1. Does the `halfway` method modify the original point `p` or `q`?**
* **Answer**: No. It calculates new values and creates a *third* point. The original points `p` and `q` remain unchanged.

**2. How does the computer know what `Point(mx, my)` is inside the method?**
* **Answer**: The method is inside the `Point` class definition, so it has access to its own class constructor to make new instances of itself.

**3. What is the "type" of the variable `mid` in the code above?**
* **Answer**: Its type is `<class '__main__.Point'>`. It is a full-featured object, not just a list or a tuple of two numbers.

### Key Logic Summary
| Operation | Logic |
| :--- | :--- |
| **Input** | `self` (Point A) and `target` (Point B) |
| **Process** | Arithmetic average of $x$ and $y$ |
| **Output** | A **new** Instance of the `Point` class |

# 17.10. Glossary

This glossary defines the essential vocabulary for Object-Oriented Programming (OOP). These terms represent the shift from writing isolated functions to building integrated, data-driven "objects."

| Term | Definition |
| :--- | :--- |
| **Attribute** | A named data item (variable) stored within an instance (e.g., `self.x`). |
| **Class** | A blueprint or template that defines a new data type and its behaviors. |
| **Constructor** | The "factory" function (named after the class) used to create new instances. |
| **Initializer Method** | The `__init__` method that automatically sets the object's starting state. |
| **Instance / Object** | A specific realization of a class. These terms are used interchangeably. |
| **Instantiate** | The act of creating a new object and running its initializer. |
| **Method** | A function defined inside a class that is called on an instance (e.g., `p.getX()`). |
| **OOP** | A programming paradigm that organizes code into classes and methods. |
| **Object-Oriented Language** | A language (like Python) that supports classes, methods, and inheritance. |



### Visualizing the Concepts
To keep these straight, think of a **Class** as a cookie cutter, and the **Instances** as the individual cookies. The **Initializer** is the process of adding the initial toppings (attributes) to each cookie as it's made.

````python
# Demonstrating the Glossary Terms

# 1. Class: The Blueprint
class Robot:
    """A class representing a simple robot."""

    # 2. Initializer Method: Sets up the State
    def __init__(self, name, model):
        # 3. Attributes: The data items
        self.name = name
        self.model = model
        self.battery = 100

    # 4. Method: A function inside a class
    def charge(self):
        self.battery = 100
        print(f"{self.name} is now fully charged!")

# 5. Instantiation: Using the Constructor to create an Instance
bot1 = Robot("Robbie", "R2") # bot1 is an Object (Instance)

# Using a Method
bot1.charge()

# Accessing an Attribute
print(f"Robot Name: {bot1.name}")
````
