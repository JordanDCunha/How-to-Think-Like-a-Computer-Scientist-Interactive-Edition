# Chapter 3.2: How to Avoid Debugging

The best way to debug is to write code in a way that prevents bugs from happening in the first place. This chapter introduces a professional workflow used in the industry called **Incremental Development**.

### The Three Pillars of Clean Coding:
1.  **Understand the Problem:** You don't need the whole solution yet, but you must know what the output should look like for a specific input.
2.  **Start Small:** Write 2–3 lines. Run them. If they work, you have a solid foundation.
3.  **Keep Improving (Iterate):** Add one small feature at a time. If the program breaks, the bug is almost certainly in the last 2 lines you just wrote.



> **Note:** This "Start Small and Improve" mindset is the basis of **Agile Software Development**, a methodology used by major tech companies like Google and Amazon.

```python
# Step 1: Start Small - Just get the input and verify it works
current_time = input("What is the current time (in hours 0-23)?")
wait_time = input("How many hours do you want to wait? ")

print("Current:", current_time)
print("Wait:", wait_time)
```

### Identifying the "String Trap"
In the next step, if we try to add these variables, we run into a common error. Because `input()` returns strings, Python performs **concatenation** (sticking them together) instead of math.

```python
# Step 2: The wrong way (Concatenation Error)
# If current is 8 and wait is 5, this prints 85, not 13!
final_time = current_time + wait_time
print("Incorrect Addition (Strings):", final_time)

# Step 3: The right way (Conversion)
current_time_int = int(current_time)
wait_time_int = int(wait_time)
final_time_int = current_time_int + wait_time_int
print("Correct Addition (Integers):", final_time_int)
```

### Testing Boundary Conditions
A program that works for `8 + 5 = 13` might fail for `17 + 9 = 26`, because a clock only goes up to 23. To fix this, we use the **Modulus Operator (%)** to wrap the time back around.

**Testing checklist:**
* Test "Middle" values (e.g., 8 and 5).
* Test "Boundaries" (e.g., current time 23, wait time 1).
* Test "Zero" (e.g., wait time 0).

```python
# Step 4: Final Polished Version
current_time_int = int(input("What is the current time (0-23)? "))
wait_time_int = int(input("How many hours to wait? "))

# Use % 24 to handle the wrap-around for a 24-hour clock
final_time = (current_time_int + wait_time_int) % 24

print("The time after waiting is:", final_time)
```

### Check Your Understanding

**Question:** What happens if you add two inputs without converting them to `int`?
**Answer:** Python performs **string concatenation**. If you input `10` and `5`, the result will be `"105"` instead of `15`.

**Question:** Why use `% 24`?
**Answer:** It ensures that if the total hours exceed 23 (the end of the day), the clock starts over at 0 (midnight).

# Chapter 3.3: Beginning Tips for Debugging

Writing code and debugging code are two different skills. If writing code is like being an architect, **debugging is like being a detective.** You have to look at the evidence to find the "criminal" (the bug) hiding in your logic.

### Rule #1: Everyone is a suspect (Except Python)
It is very tempting to think, "Python is broken!" or "The computer is wrong!" However, Python is a very mature language used by millions. It executes exactly what you tell it to do.
* **The "Detective" Mindset:** Assume the bug is in your instructions, not the computer's brain.

### Rule #2: Find Clues
A detective doesn't guess; they look for evidence. In Python, you have two primary ways to find clues:
1.  **Error Messages:** These are Python's way of telling you *exactly* where it got confused.
2.  **Print Statements:** These let you "peek" inside the computer's memory while the program is running to see what values your variables actually hold.

### Using the "Print Detective" Method
If your math is coming out wrong, don't just stare at the code. Print the values at every step. This reveals exactly where the logic diverges from your expectations.

```python
# Activity 3.3.1: The "Print Detective" in Action

# Problem: We want to calculate the total price of 3 shirts at $20 each
# with a $5 discount.
price = 20
quantity = 3
discount = 5

# Intent: (20 * 3) - 5 = 55
# Bug: Forgot parentheses or order of operations
total = price * quantity - discount

# If the result was confusing, we would "Find Clues" like this:
print("CLUE - Price:", price)
print("CLUE - Quantity:", quantity)
print("CLUE - Subtotal (Price * Quantity):", price * quantity)
print("CLUE - Final Total:", total)

# This confirms the math is working as written, helping us rule out suspects!
```

### Summary Checklist for Debugging:
* **Don't panic:** Errors are just feedback, not a failure.
* **Read the Error:** Does it say `NameError`? (Check your spelling). Does it say `TypeError`? (Check your `int()` conversions).
* **Verify your Suspects:** Use `print()` to make sure your variables hold the values you *think* they do.

# Chapter 3.4: Know Your Error Messages

Learning to read error messages is like learning a new language. These messages aren't just telling you that you failed; they are giving you specific clues about where the "crime" occurred.

### Common Beginner Errors
Statistically, nearly 90% of the errors you will face as a beginner fall into four categories:
1. **ParseError (SyntaxError):** Your "grammar" is wrong (missing commas, quotes, or parentheses).
2. **TypeError:** You are trying to combine incompatible types (like adding a string to an integer).
3. **NameError:** You used a variable name that hasn't been defined or is spelled incorrectly.
4. **ValueError:** You passed the right *type* of data but the wrong *value* (like trying to turn the string "hello" into an integer).

### 1. ParseError (SyntaxError)
This happens when Python can't "parse" your code because it doesn't follow the rules of the language.
* **The Punctuation Trap:** Usually caused by a missing `)`, `"`, or `:`.
* **Detective Tip:** If line 5 has a ParseError, look at line 4. Often, a missing parenthesis on the previous line causes the error to show up on the next one!

```python
# Activity 3.4.1.1: Spot the ParseError
# Can you see the missing parenthesis?
current_time_str = input("What is the current time (0-23)?")
wait_time_str = input("How many hours to wait?" # Missing ')' here!

current_time_int = int(current_time_str)
print("This won't run until the line above is fixed.")
```

### 2. TypeError and NameError
* **TypeError:** Common when you forget to convert `input()` strings to `int()`.
* **NameError:** Usually a typo. Remember that `wait_time` and `wait_time_int` are different suspects!

**Detective Tip for NameErrors:** Use "Find" (Ctrl+F) in your browser/editor to search for the variable name in the error message. If it only appears once, or if the spelling changes slightly, you've found your bug.

```python
# Activity 3.4.3.1: Spot the NameError
str_time = input("What time is it now?")
time = int(str_time)

# Typo below: 'wai_time' vs 'wait_time'
wai_time = int(input("Wait time? "))

# This will trigger a NameError because 'wait_time' was never defined
# final = time + wait_time
```

### 3. ValueError
This happens when the code is technically correct, but the *data* provided is bad.
* **Example:** `int("ten")` or `int("")`.
* Python knows how to turn a string into an int, but it doesn't know how to turn the word "ten" into the number `10`.

# Chapter 3.5: Summary of Debugging

This chapter focused on shifting your mindset from a "coder" to a "detective." Before we move on to the fun visual parts of Python, let's lock in these three golden rules of debugging.

### The Three Golden Rules:
1.  **Read the Error Messages:** Don't treat them as a failure message. Treat them as a GPS coordinate. They tell you the **Type** of error and the **Location** (line number).
2.  **`print` Statements are Your Friends:** When you aren't sure why a calculation is wrong, use `print()` to reveal the values and types of your variables at every step.
3.  **Work Backward:** Because Python executes from top to bottom, an error on Line 10 is often caused by a mistake on Line 5. If a variable is "Undefined," look up!

### Debugging Strategy Checklist
If your program isn't working, follow this detective's checklist:
* **Step 1:** Did I spell all my variables exactly the same way? (Case matters!)
* **Step 2:** Did I convert my inputs to `int` or `float` before doing math?
* **Step 3:** Did I close all my parentheses `()` and quotes `""`?
* **Step 4:** If I'm still stuck, have I added `print()` statements to check my data?

```python
# A "Detective" template for future projects:
# If this code failed, we use prints to find the "suspect"

try:
    # Let's assume the user enters '10'
    val_str = input("Enter a number: ")
    print("DEBUG: val_str is", val_str, "type is", type(val_str))

    val_int = int(val_str)
    print("DEBUG: val_int is", val_int, "type is", type(val_int))

    result = val_int * 2
    print("Final Result:", result)

except Exception as e:
    print("DETECTIVE REPORT: We found a bug!")
    print("Error Type:", type(e).__name__)
    print("Error Details:", e)
```
