# 6.1. Functions

In Python, a **function** is a named sequence of statements that belong together. Their primary purpose is to help us organize programs into chunks that match how we think about the solution to the problem.

### The Syntax
The syntax for a **function definition** is:

```python
def name( parameters ):
    statements

### Key Rules for Definitions:
* **Naming**: Use any names except Python keywords. Names must follow legal identifier rules.
* **Parameters**: Specify what information the function needs to do its work.
* **Header & Body**: The header line begins with a keyword (`def`) and ends with a colon (`:`). The body consists of one or more statements indented by **4 spaces**.

### Formal Parameters vs. Arguments
* **Formal Parameters**: The list of names in the definition describing what the function needs to receive.
* **Arguments (Actual Parameters)**: The values passed to the function by the user when invoking it.

### Abstraction and "Black-Boxes"
A function is an **abstraction**. For example, "Draw a square" hides the smaller steps of moving and turning. This is often called a **black-box diagram** because the user knows the name and arguments required, but the internal details are hidden.

### Docstrings
If the first thing after the function header is a string, it is a **docstring**.
* Triple-quoted strings (`"""`) are standard.
* They are retrievable at runtime using `function_name.__doc__`.

### Function Invocation (Calls)
Defining a function does **not** run it. You must **call** (invoke) the function using its name and parentheses.

> **⚠️ Warning:** Parentheses `()` are **required** for a call. Without them, you are only referring to the function object, not executing it.

````python
import turtle

def drawSquare(t, sz):
    """Make turtle t draw a square of with side sz."""
    for i in range(4):
        t.forward(sz)
        t.left(90)

wn = turtle.Screen()
wn.bgcolor("lightgreen")

alex = turtle.Turtle()
drawSquare(alex, 50) # First invocation

alex.penup()
alex.goto(100,100)
alex.pendown()

drawSquare(alex, 75) # Second invocation
wn.exitonclick()
````

````python
import turtle

def drawMulticolorSquare(t, sz):
    """Make turtle t draw a multi-colour square of sz."""
    for i in ['red','purple','hotpink','blue']:
        t.color(i)
        t.forward(sz)
        t.left(90)

wn = turtle.Screen()
wn.bgcolor("lightgreen")

tess = turtle.Turtle()
tess.pensize(3)

size = 20
for i in range(15):
    drawMulticolorSquare(tess, size)
    size = size + 10
    tess.forward(10)
    tess.right(18)

wn.exitonclick()
````

# 6.2. Functions that Return Values

Most functions require **arguments**—values that control how the function does its job. For example, Python's built-in `abs` function requires one number to compute its absolute value, while `pow` requires two (base and exponent).

### Fruitful vs. Non-fruitful Functions
* **Fruitful Functions**: Functions that return a value after execution (e.g., `abs`, `int`, `pow`, `range`). The return value can be used to build complex expressions.
* **Non-fruitful Functions (Procedures)**: Functions executed for their "side effects," such as `drawSquare`, which causes a turtle to move but does not result in a data value for the program to use later.

````python
import math

# Functions with different numbers of arguments
print("Absolute value:", abs(-5))
print("Power (2^3):", math.pow(2, 3))

# max() can take any number of arguments or a list
print("Max of multiple numbers:", max(33, 125, 503, 1))
````

### Common Mistake: Print vs. Return
A frequent error is using `print()` inside a function instead of `return`.

* `print()` shows a value to the human user but does not give the value to the program.
* `return` gives the value to the program so it can be stored in a variable or used in further calculations.

If you attempt to assign the result of a function that only uses `print()` to a variable, that variable will be `None`.

````python
def bad_square(x):
    y = x * x
    print(y)  # This shows the value, but does not "return" it

toSquare = 10
result = bad_square(toSquare)

print("The value of result is:", result) # This will print 'None'
````

# 6.3. Unit Testing

A **test case** expresses requirements for a program in a way that can be checked automatically. A **unit test** is an automated procedure used to validate that individual units of code (like a single function) are working properly. A collection of these tests is called a **test suite**.

### Why Write Test Cases?
1. **Clarity**: It forces you to be concrete about what the code should do before you write it.
2. **Automated Feedback**: Tests provide immediate notification if your logic is incorrect.
3. **Regression Testing**: In large projects, tests are run every time code changes to ensure nothing broke.

### The `assert` Statement
One way to implement unit tests in Python is with the `assert` keyword.
* **How it works**: `assert` is followed by a Python expression.
* **If True**: Nothing happens, and the program continues to the next line.
* **If False**: The interpreter raises a **runtime error**, alerting you that an assumption was wrong.

`assert` helps catch bugs early by flagging when a variable isn't what you expect it to be (e.g., expecting a list but getting an integer).

````python
# Testing assumptions about division
assert type(9 // 5) == int

# This will fail and raise an AssertionError because 9.0 // 5 is 1.0 (a float)
try:
    assert type(9.0 // 5) == int
    print("Test passed!")
except AssertionError:
    print("Test failed: 9.0 // 5 is actually a float!")
````

### Using `assert` with Loops
We can use loops to check consistency across data structures. This is useful for ensuring all items in a list meet a certain criteria, such as having the same data type.

Why doesn't `assert` print "Pass"? To avoid cluttering the output. You only want to know when something is **wrong**.

````python
# Success case: All items are strings
lst = ['a', 'b', 'c']
first_type = type(lst[0])
for item in lst:
    assert type(item) == first_type

# Failure case: The integer 17 will trigger an error
lst2 = ['a', 'b', 'c', 17]
print("Starting second test...")
first_type = type(lst2[0])
for item in lst2:
    # This will crash the program when it hits index 3
    assert type(item) == first_type
````

# 6.4. Variables and Parameters are Local

When you create a variable inside a function using an assignment statement, it is a **local variable**. It is called "local" because it only exists within that specific function. You cannot use it outside of the function's body.

### Lifetime of a Variable
A variable's **lifetime** is the period during which the variable exists in memory:
* **Creation**: Local variables (and formal parameters) are created when the function is called.
* **Destruction**: When the function completes its execution and returns, all local variables are destroyed.

Because of this, a function cannot "remember" a local variable's value from one call to the next. Every call starts fresh with new local variables.

````python
def square(x):
    y = x * x  # y is a local variable
    return y

z = square(10)

# The following line would cause a NameError because 'y' does not exist here
try:
    print(y)
except NameError as e:
    print(f"Error caught: {e}")
````

### Global Variables vs. Local Scope
Python follows a specific order when looking up variable names:
1. **Local Scope**: First, it looks at variables defined inside the current function.
2. **Global Scope**: If not found locally, it looks at variables defined at the top level of the script.

**Bad Practice Alert:** While a function *can* read a global variable, it is considered **bad form**. Functions should be self-contained; if a function needs a value, that value should be passed in as a parameter.

````python
# POOR PRACTICE: Accessing a global variable 'power'
def badsquare(x):
    y = x ** power # 'power' is not defined in the function
    return y

power = 2
result = badsquare(10)
print("Badsquare result:", result)

# BETTER PRACTICE: Pass 'power' as a parameter
def goodsquare(x, p):
    return x ** p

print("Goodsquare result:", goodsquare(10, 2))
````

### Shadowing Global Variables
If a local variable is assigned a value that shares the same name as a global variable, the local variable **shadows** the global one.
* Inside the function, Python will only "see" the local version.
* The global version remains unchanged outside the function.

### Parameters are Local
Formal parameters act exactly like local variables. Assigning a new value to a parameter inside a function will **not** change the value of the variable used as the actual argument in the global scope.

````python
def square_shadow(x):
    y = x * x
    x = 0  # Changing the local parameter x to 0
    print("Inside function local x:", x)
    return y

x = 2
result = square_shadow(x)

print("Global x after function call:", x) # x is still 2
print("Function result:", result)
````

# 6.5. The Accumulator Pattern

The **accumulator pattern** is a fundamental programming strategy where a variable is used to "accumulate" a result over time, usually within a loop.

### How it Works: Squaring by Addition
To understand this, consider multiplication as repeated addition. To square a number $n$, you can add $n$ to itself $n$ times.
* **Example**: $3^2 = 3 + 3 + 3 = 9$.

### The General Pattern
1. **Initialize**: Set an accumulator variable to an initial value (the identity element).
2. **Repeat**: Use a loop to update the variable.
3. **Return/Result**: After the loop finishes, the variable contains the final accumulated value.

````python
def square(x):
    # 1. Initialize the accumulator
    runningtotal = 0

    # 2. Repeat x times
    for counter in range(x):
        # 3. Update the accumulator
        runningtotal = runningtotal + x

    # 4. Return the result
    return runningtotal

toSquare = 10
squareResult = square(toSquare)
print("The result of", toSquare, "squared is", squareResult)
````

### Critical Success Factors
* **Initialization Location**: The accumulator must be initialized **before** the loop. If you initialize it inside the loop, the value will reset to 0 every single time the loop runs, losing all previous progress.
* **The Identity Element**:
    * If adding (Accumulating a sum), initialize to **0** (Additive Identity).
    * If multiplying (Accumulating a product), initialize to **1** (Multiplicative Identity).

````python
# Experimenting with the pattern
def sum_odd_numbers(n):
    """Adds up the first n odd numbers."""
    total = 0
    current_odd = 1
    for i in range(n):
        total = total + current_odd
        current_odd = current_odd + 2
    return total

n = 5
print(f"Sum of the first {n} odd numbers is: {sum_odd_numbers(n)}")
````

### Variation and Debugging
When modifying the accumulator pattern, you must ensure:
1. **The loop repeats the correct number of times**: Check your `range()` carefully.
2. **Handle different inputs**: Our addition-based `square` function might fail with negative numbers because `range(negative_number)` doesn't execute the loop body. This is a **semantic error** where the code runs but gives the wrong logic.

# 6.6. Functions can Call Other Functions

One of the most powerful features of functions is that they can call other functions. This is the basis for **functional decomposition**: the process of breaking a large, complex problem into smaller, manageable subproblems.

### Functional Decomposition Example
In the example below, `sum_of_squares` doesn't need to know how to calculate a square; it simply delegates that task to the `square` function.



### Key Observations:
* **Independence**: Each function call creates its own local scope. Even if two functions use the same variable name (like `x`), they are completely separate and do not interfere with each other.
* **Flow of Control**: When `sum_of_squares` calls `square`, the program execution jumps to the `square` function, completes it, and then returns to the exact spot in `sum_of_squares` where it left off.

````python
def square(x):
    y = x * x
    return y

def sum_of_squares(x, y, z):
    # Calling the 'square' function three times
    a = square(x)
    b = square(y)
    c = square(z)

    return a + b + c

# Global scope
a = -5
b = 2
c = 10
result = sum_of_squares(a, b, c)
print(f"The sum of squares is: {result}")
````

### Generalization
**Generalization** is a problem-solving technique where you look for patterns and relationships to create a more flexible solution.

For example, a **square** is just a special case of a **rectangle** where the width and height are equal. By writing a general `drawRectangle` function first, we can write a much simpler `drawSquare` function that just calls the rectangle function with the same value for both sides.

````python
import turtle

def drawRectangle(t, w, h):
    """Get turtle t to draw a rectangle of width w and height h."""
    for i in range(2):
        t.forward(w)
        t.left(90)
        t.forward(h)
        t.left(90)

def drawSquare(tx, sz):
    """A version of drawSquare that uses generalization."""
    # We call the existing rectangle function with equal width and height
    drawRectangle(tx, sz, sz)

# Setup
wn = turtle.Screen()
wn.bgcolor("lightgreen")
tess = turtle.Turtle()

# Execute
drawSquare(tess, 50)

wn.exitonclick()
````

### Why is this worth the trouble?
1. **Naming and Abstraction**: It allows you to name a group of statements, hiding complex math or logic behind a single, descriptive command.
2. **Reducing Repetition**: It eliminates redundant code. Instead of writing the "draw" logic twice, you write it once and reuse it.
3. **Mental Chunking**: It matches how humans think about problems. We think "draw a square," not "move forward, turn 90 degrees, repeat four times."

# 6.7. Flow of Execution Summary

Understanding the **flow of execution** is critical when working with functions. This refers to the order in which Python executes the statements in your program.

### Basic Rules of Flow
1. **Top to Bottom**: Execution always begins at the first statement of the program and moves downward, one statement at a time.
2. **Definitions vs. Execution**: `def` statements do not cause the function body to run immediately. They simply tell Python the function exists. The code inside is skipped until the function is called.
3. **The Detour**: A function call is like a detour. The program "jumps" to the function's first line, runs every statement in that block, and then "jumps back" to the exact spot where it left off.



### Nested Calls
Python is excellent at tracking its location, even when functions call other functions (nested calls). It maintains a "stack" of where to return to next. Once a function completes, the program picks up in the function that called it, eventually returning to the main program level.

> **Pro Tip:** When reading code, don't just read top-to-bottom like a book. Follow the execution. Skip function bodies until you see them being called.

````python
# Follow the execution order here:
def pow(b, p):
    # This runs 4th (when called by square)
    y = b ** p
    return y

def square(x):
    # This runs 2nd (when called by main)
    # Then it calls pow, so it pauses here...
    a = pow(x, 2)
    # This runs 5th (after pow returns)
    return a

# --- MAIN PROGRAM ---
# 1. This runs first
n = 5
# 2. This runs second, jumping to the square function
result = square(n)
# 3. This runs last, after all functions return
print("The result is:", result)
````

### Advanced Topic: `if __name__ == "__main__":`

Python defines a special variable called `__name__`.
* When you run a script **directly**, Python sets `__name__` to `"__main__"`.
* If you **import** the script into another file as a module, `__name__` is set to the file's name instead.

### Why use this check?
This allows you to include code (like tests or a demonstration) that only runs when the file is executed by itself. If someone else imports your file to use your functions, your `main()` function won't automatically run and interfere with their program.

````python
def squareit(n):
    return n * n

def cubeit(n):
    return n * n * n

def main():
    # This logic only runs if the script is executed directly
    anum = 5
    print(f"{anum} squared is {squareit(anum)}")
    print(f"{anum} cubed is {cubeit(anum)}")

# Conditional execution check
if __name__ == "__main__":
    main()
````

# 6.10. Composition

As we have seen, you can call one function from within another. This ability to build complex functions by using other, simpler functions is called **composition**.

### Example: Area of a Circle from Two Points
Suppose you want to find the area of a circle, but you are only given the coordinates for the **center point** $(xc, yc)$ and a **point on the perimeter** $(xp, yp)$.

To solve this, we can compose our existing logic:
1.  **Find the Radius**: The radius is simply the distance between the center and the perimeter point. We can use our `distance` function for this.
2.  **Calculate the Area**: Use the radius with our `area` function ($Area = \pi r^2$).



### Why use Composition?
* **Reusability**: We don't have to rewrite the distance formula or the area formula.
* **Readability**: The code reads more like a logical sequence of steps.
* **Maintainability**: If we find a better way to calculate distance, we only have to change it in one place (the `distance` function).

````python
def distance(x1, y1, x2, y2):
    """Calculates the distance between two points."""
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    return dsquared**0.5

def area(radius):
    """Calculates the area of a circle given its radius."""
    return 3.14159 * radius**2

def area_from_points(xc, yc, xp, yp):
    """
    Composes distance and area functions to find the
    area of a circle defined by two points.
    """
    # Step 1: Use the distance function to find the radius
    r = distance(xc, yc, xp, yp)

    # Step 2: Use the area function with that radius
    result = area(r)

    return result

# Test: Center at (0,0), Perimeter at (1,1)
# Radius should be approx 1.414, Area should be approx 6.28
print(f"Area of the circle: {area_from_points(0, 0, 1, 1)}")
````

### Concise Composition

# 6.11. A Turtle Bar Chart

By combining functions, loops, and some new turtle methods, we can create a program to visualize data in a **bar chart**.

### New Turtle Methods
* **`t.write("text")`**: Displays a string of text at the turtle's current position.
* **`t.begin_fill()`** and **`t.end_fill()`**: These wrap around the code that draws a shape to fill it with color.
* **`t.fillcolor("color")`**: Sets the internal color used for filling shapes, which can be different from the pen color.
* **`wn.setworldcoordinates()`**: Rescales the window. This allows us to set the bottom-left corner to (0,0) and the top-right corner to match our data dimensions.



### Functional Decomposition
To solve this, we use **mental chunking**:
1.  **The Chunk**: Create a function `drawBar` that draws exactly one vertical rectangle of a given height.
2.  **The Iteration**: Use a `for` loop to iterate through a list of data, calling `drawBar` for each value.

````python
import turtle

def drawBar(t, height):
    """ Get turtle t to draw one bar, of height. """
    t.begin_fill()               # Start filling the bar
    t.left(90)                   # Face up
    t.forward(height)            # Draw the left side

    # Write the value above the bar
    t.write("  " + str(height))

    t.right(90)
    t.forward(40)                # Draw the top (width = 40)
    t.right(90)
    t.forward(height)            # Draw the right side
    t.left(90)                   # Face right again, ready for the next bar
    t.end_fill()                 # Fill the shape

def main():
    # Data to be charted
    xs = [48, 117, 200, 240, 160, 260, 220]
    maxheight = max(xs)
    numbars = len(xs)
    border = 10

    wn = turtle.Screen()
    # Rescale the window to fit the bars perfectly
    wn.setworldcoordinates(0-border, 0-border, 40*numbars+border, maxheight+border)
    wn.bgcolor("lightgreen")

    tess = turtle.Turtle()
    tess.color("blue")           # Pen color
    tess.fillcolor("red")        # Bar fill color
    tess.pensize(3)

    for val in xs:
        drawBar(tess, val)

    wn.exitonclick()

if __name__ == "__main__":
    main()
````

### Design Considerations
* **Coordinate Scaling**: By using `setworldcoordinates`, we avoid having the bars start in the middle of the screen. We calculate the width of the window based on the number of bars ($40 \times \text{number of bars}$) and the height based on the `max()` value in our list.
* **Type Conversion**: Since `t.write()` requires a string, we must use `str(height)` to convert our numerical data.
* **Refinement**: In the code above, if the label is covered by the bar line, you can lift the pen (`t.penup()`), move slightly higher, write the text, and move back before finishing the fill.

## Check Your Understanding

**1. What is the purpose of `t.begin_fill()` and `t.end_fill()`?**
* **Answer:** They are used to fill a closed shape with the color specified by `t.fillcolor()`. Everything drawn between these two calls becomes the boundary of the filled area.

**2. How do we ensure the window is large enough to show the tallest bar?**
* **Answer:** We find the maximum value in our dataset using `max(xs)` and use that value to set the upper vertical boundary in `setworldcoordinates`.

**3. Why do we turn the turtle 90 degrees left at the start of `drawBar`?**
* **Answer:** Turtles typically start facing East (right). To draw a vertical bar, we must first turn the turtle to face North (up).

# 6.12. Glossary

This section provides a summary of the key terms introduced in this chapter regarding function logic and program development.

### Function Types and Returns
* **fruitful function**: A function that yields a return value instead of `None`.
* **return value**: The value provided as the result of a function call.
* **None**: A special Python value returned by functions that do not execute a `return` statement with a specific argument.
* **chatterbox function**: A function which interacts with the user (using `input` or `print`) when it should not. **Silent functions** that simply convert input arguments into output results are generally more useful and reusable.

### Program Structure and Logic
* **composition (of functions)**: Calling one function from within the body of another, or using the return value of one function as an argument to the call of another.
* **dead code**: Part of a program that can never be executed, often because it appears after a `return` statement.
* **temporary variable**: A variable used to store an intermediate value in a complex calculation, improving readability and debugging.



### Development Techniques
* **incremental development**: A program development plan intended to simplify debugging by adding and testing only a small amount of code at a time.
* **scaffolding**: Code that is used during program development to assist with development and debugging. The unit test code (like `assert` statements) added in this chapter are examples of scaffolding.

````python
# Demonstrating Glossary Terms

def add_two(n):
    return n + 2

def multiply_by_three(n):
    # COMPOSITION: Using the return of add_two as an argument
    return add_two(n) * 3

def example_dead_code():
    return "I am fruitful"
    print("I am dead code") # This will never execute

# SCAFFOLDING / UNIT TEST
assert multiply_by_three(2) == 12
print("Test Passed!")

# CHATTERBOX vs SILENT
# This is a chatterbox because it prints instead of returning
def chatterbox(n):
    print(n + 1)

# This is silent and more useful
def silent(n):
    return n + 1
````
