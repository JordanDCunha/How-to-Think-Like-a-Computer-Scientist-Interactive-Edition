# 7.1. Boolean Values and Boolean Expressions

The Python type for storing true and false values is called **bool**, named after the British mathematician George Boole. Boolean Algebra is the basis of all modern computer arithmetic.

### The Boolean Values
There are only two values in the `bool` data type:
* **True**
* **False**

**Important Rules:**
1.  **Capitalization**: Python is case-sensitive. You must use `True` and `False`. `true` or `false` (lowercase) will result in an error.
2.  **Not Strings**: Boolean values are not surrounded by quotes. `"True"` is a string, but `True` is a boolean.

````python
# Checking Boolean types
print(f"Type of True: {type(True)}")
print(f"Type of 'True' (with quotes): {type('True')}")

# Comparing values
print(f"Is 5 equal to 5? {5 == 5}")
print(f"Is 5 equal to 6? {5 == 6}")

j = "hel"
print(f"Does 'hel' + 'lo' equal 'hello'? {j + 'lo' == 'hello'}")
````

### Comparison Operators
A **boolean expression** is an expression that evaluates to a boolean value. The most common way to create one is using comparison operators:

| Operator | Meaning |
| :--- | :--- |
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `> =` | greater than or equal to |
| `<= ` | less than or equal to |



### Common Pitfalls
* **Assignment vs. Comparison**: `=` is used to assign a value to a variable. `==` is used to compare two values.
* **Symbol Order**: Operators like `>=` and `<=` must have the equal sign second. `=>` or `=<` are syntax errors.
* **Symmetry**: While `a == 7` is the same as `7 == a`, the assignment `a = 7` is legal while `7 = a` is not because you cannot assign a value to a literal number.

# 7.2. Logical Operators

Logical operators allow us to build complex decisions by combining multiple boolean expressions. Python has three main logical operators: `and`, `or`, and `not`.

### The Three Operators
1.  **`and`**: True only if **both** sides are true. (e.g., `x > 0 and x < 10`)
2.  **`or`**: True if **at least one** side is true.
3.  **`not`**: Negates the expression (True becomes False, False becomes True).

### Truth Tables
A **truth table** is used to show every possible outcome of a logical operation based on the inputs ($a$ and $b$).

| a | b | a and b | a or b | not a |
| :--- | :--- | :--- | :--- | :--- |
| True | True | **True** | **True** | False |
| True | False | False | **True** | False |
| False | True | False | **True** | **True** |
| False | False | False | False | **True** |



[Image of a truth table for AND, OR, and NOT logical gates]


### ⚠️ Common Pitfall: The "Shortcut" Error
In English, we say "If the number is 5, 6, or 7." In Python, you **cannot** write `if number == 5 or 6 or 7`.
* **Incorrect**: `number == 5 or 6 or 7` (This actually evaluates as `(number == 5) or (6) or (7)`, and since 6 is non-zero, it is always True!)
* **Correct**: `number == 5 or number == 6 or number == 7`

````python
x = 5
print(f"Is 0 < {x} < 10? {x > 0 and x < 10}")

n = 25
# Check if divisible by 2 OR 3
print(f"Is {n} divisible by 2 or 3? {n % 2 == 0 or n % 3 == 0}")

# Demonstrating the 'not' operator
is_sunny = False
print(f"Is it NOT sunny? {not is_sunny}")
````

### Logical Opposites and Simplification
Every comparison operator has a logical opposite. Using these can make your code much easier to read by removing unnecessary `not` operators.

| Operator | Opposite |
| :--- | :--- |
| `==` | `!=` |
| `>` | `<=` |
| `<` | `>=` |



### De Morgan's Laws
These laws help simplify complex negations:
1.  `not (x and y)` is the same as `(not x) or (not y)`
2.  `not (x or y)` is the same as `(not x) and (not y)`

**Example**:
* *Before*: `if not (battery > 0.5 and storage > 0.15):`
* *After*: `if battery <= 0.5 or storage <= 0.15:`

````python
# Phone Update Example
battery = 0.45
storage = 0.20

# Approach 1: Using 'not' and 'and'
if not (battery >= 0.50 and storage >= 0.15):
    print("Update Failed: Condition not met.")

# Approach 2: Using logical opposites (Clearer!)
if battery < 0.50 or storage < 0.15:
    print("Update Failed: Low battery or storage.")
else:
    print("Updating now...")
````

# 7.3. Precedence of Operators

When a single expression contains multiple operators, Python needs a set of rules to decide which parts to evaluate first. This is known as **operator precedence**.

### The Order of Operations
Python follows a strict hierarchy. Generally, it processes math first, then comparisons, and finally logic:
1.  **Arithmetic**: Calculations like exponents (`**`) happen first, followed by multiplication/division, then addition/subtraction.
2.  **Relational**: Comparison operators (`==`, `>`, `<=`, etc.) are evaluated next.
3.  **Logical**: The logic gates `not`, `and`, and finally `or` are processed last.



### Why It Matters
In the expression `x * 5 >= 10 and y - 6 <= 20`, Python doesn't need parentheses to know what to do. It will:
* Calculate `x * 5` and `y - 6`.
* Compare those results to `10` and `20`.
* Finally, combine those two Boolean results with the `and`.

### Precedence Table (High to Low)

| Level | Category | Operators |
| :--- | :--- | :--- |
| 7 (Highest) | Exponent | `**` |
| 6 | Multiplicative | `*`, `/`, `//`, `%` |
| 5 | Additive | `+`, `-` |
| 4 | Relational | `==`, `!=`, `<=`, `>=`, `>`, `<` |
| 3 | Logical NOT | `not` |
| 2 | Logical AND | `and` |
| 1 (Lowest) | Logical OR | `or` |

````python
# Demonstrating Precedence
# Expression: 5 * 3 > 10 and 4 + 6 == 11

# Step 1: Arithmetic (5*3=15, 4+6=10)
# Step 2: Relational (15 > 10 is True, 10 == 11 is False)
# Step 3: Logical (True and False is False)

result = 5 * 3 > 10 and 4 + 6 == 11
print(f"The result of the expression is: {result}")

# Using parentheses to show the order clearly (Same result)
explicit_result = ((5 * 3) > 10) and ((4 + 6) == 11)
print(f"The explicit result is: {explicit_result}")
````

# 7.4. Conditional Execution: Binary Selection

To write useful programs, we need the ability to check conditions and change the program's behavior. This is called **selection** or **conditional execution**.

### Binary Selection (if-else)
The most common form of selection is the `if-else` statement. It is called **binary selection** because it creates two possible paths (branches) for the flow of execution.



### Syntax and Structure
An `if` statement is a **compound statement**, meaning it has a header and a body:
1.  **Header**: Starts with `if`, followed by a **boolean expression** (the condition), and ends with a colon `:`.
2.  **The Block**: The indented statements following the header.
3.  **The Else Clause**: Provides an alternative path if the condition is `False`. It must be at the same indentation level as the `if`.

```python
if CONDITION:
    # This block executes if CONDITION is True
    STATEMENTS_1
else:
    # This block executes if CONDITION is False
    STATEMENTS_2

````python
# Example: Checking for even or odd numbers
x = 15

if x % 2 == 0:
    # Executed if the remainder is 0
    print(x, "is even")
else:
    # Executed if the remainder is NOT 0
    print(x, "is odd")

# This statement is outside the if-else block and always runs
print("Processing complete.")
````

### Key Rules for Indentation
* The statements inside the `if` or `else` blocks must be **indented** (usually 4 spaces).
* There is no limit to the number of statements in a block, but there must be **at least one**.
* The `else` keyword is always at the same indentation level as its corresponding `if`.
* Any code that is **unindented** back to the level of the `if` header marks the end of the conditional structure and will execute normally regardless of the condition.

# 7.5. Omitting the else Clause: Unary Selection

Sometimes you only want to execute code if a condition is **True**, but you don't need to do anything special if it is **False**. This is called **unary selection**.

### How Unary Selection Works
In this form, the `else` clause is omitted entirely.
* If the condition is **True**: The indented block runs, then the program continues to the next unindented line.
* If the condition is **False**: The indented block is skipped, and the program jumps straight to the next unindented line.



### Use Cases
Unary selection is often used for:
* **Input Validation**: Checking if a number is valid before proceeding.
* **Optional Flags**: Triggering a specific behavior only when a certain requirement is met.
* **Error Handling**: Printing a warning message only when something goes wrong.

````python
x = -10

# Unary selection: only act if the number is negative
if x < 0:
    print("The negative number", x, "is not valid here.")
    # You could also reset x here: x = 0

# This line is NOT indented, so it is NOT part of the if-statement.
# It will run regardless of whether x is positive or negative.
print("This is always printed")
````

### Indentation is Everything
In Python, the end of an `if` block is determined solely by **indentation**. As soon as you write a line of code that is "lined up" with the `if` keyword, the conditional logic is over.

> **Important**: You can have multiple `if` statements in a row without `else` clauses. Each one will be evaluated independently as the program flows from top to bottom.

# 7.6. Nested Conditionals

A **nested conditional** occurs when one conditional statement is placed inside the branch of another. This allows you to handle complex scenarios that require multiple layers of decision-making.

### How Nesting Works
In the example of comparing two numbers $x$ and $y$, there are three possibilities:
1. $x < y$
2. $x > y$
3. $x = y$

To handle this, we first check if $x < y$. If that is false, we enter the `else` block and immediately perform a **second** check to see if $x > y$.



### The Importance of Indentation
In many languages, it's hard to tell which `else` goes with which `if`. In Python, **indentation** makes it clear:
* The second `if` is indented inside the first `else`.
* The second `else` is indented even further to align with the second `if`.

> **Note:** While nesting is powerful, having too many levels of nesting can make code "messy" and hard to read. In the next section, we will look at `elif`, which is often a cleaner alternative.

````python
x = 10
y = 10

if x < y:
    # Outer block: Branch 1
    print("x is less than y")
else:
    # Outer block: Branch 2 (This contains the nested if)
    if x > y:
        # Inner block: Branch A
        print("x is greater than y")
    else:
        # Inner block: Branch B
        print("x and y must be equal")

# Try changing x to 5 or 15 to see different paths!
````

### Visualizing the Path
If $x$ and $y$ are both 10:
1.  Python checks `x < y` (10 < 10). This is **False**.
2.  It jumps to the `else` block on line 6.
3.  Inside that block, it finds a new instruction: `if x > y`.
4.  It checks `x > y` (10 > 10). This is also **False**.
5.  It jumps to the inner `else` on line 9.
6.  It executes line 10: `print("x and y must be equal")`.

# 7.7. Chained Conditionals

As programs grow, nested `if-else` statements can become deeply indented and difficult to read. Python provides the `elif` keyword (short for **else if**) to create **chained conditionals**, allowing for a "flatter" and cleaner structure.

### How it Works
A chained conditional allows you to check multiple conditions in sequence.
* Exactly **one** branch will be executed.
* Each condition is checked in order from top to bottom.
* As soon as one condition evaluates to `True`, its block runs, and Python **skips the rest** of the chain.



### Rules for Chained Conditionals
1.  **Starts with `if`**: Every chain must begin with exactly one `if`.
2.  **Middle `elif`**: You can have zero or many `elif` statements in the middle.
3.  **Final `else`**: You can have zero or one `else` at the very end. It acts as a "catch-all" if every previous condition was `False`.

````python
x = 10
y = 10

# Chained version of the comparison logic
if x < y:
    print("x is less than y")
elif x > y:
    print("x is greater than y")
else:
    print("x and y must be equal")

# Example with more than three branches
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Your grade is: {grade}")
````

### Order Matters
The sequence of conditions is critical. Even if multiple conditions are technically true, only the **first** one that Python encounters will execute.

For example, in the grade code above, a score of `95` is technically `>= 90`, `>= 80`, and `>= 70`. However, because `score >= 90` comes first, the program assigns an "A" and ignores the other `elif` branches.

# 7.8. Boolean Functions

Functions can return any type of value, including Booleans. A **Boolean function** is a function that returns either `True` or `False`. These are incredibly useful for hiding complex logic or calculations behind a simple, descriptive name.

### Defining Boolean Functions
It is a common convention to name Boolean functions like questions (e.g., `isDivisible`, `has_permission`, `is_even`).

**Concise Implementation:**
While you can use an `if-else` block inside the function to return `True` or `False`, it is more professional to return the result of a comparison directly.



| Verbose Method | Concise Method |
| :--- | :--- |
| `if x % y == 0:`<br>`    return True`<br>`else:`<br>`    return False` | `return x % y == 0` |

### Using Boolean Functions
When using these functions in an `if` statement, you do not need to compare them to `True`. The function call itself evaluates to a Boolean value.

* **Good**: `if isDivisible(10, 5):`
* **Redundant**: `if isDivisible(10, 5) == True:`

````python
def isDivisible(x, y):
    """Returns True if x is evenly divisible by y, False otherwise."""
    return x % y == 0

# The function call is used as the 'condition' for the if statement
if isDivisible(10, 5):
    print("10 is divisible by 5")
else:
    print("Those values are no good")

# Using 'not' to check for the False case
if not isDivisible(10, 3):
    print("10 is not divisible by 3")
````

### Unit Testing: Output Equivalence Classes
When testing Boolean functions, you must test for **Output Equivalence Classes**. This simply means you need at least one test case that results in `True` and at least one that results in `False`. This ensures your logic works for both possible outcomes.



### Example Test Scaffolding
Using `assert` statements (or a `test` module) allows you to verify both classes:
* `isDivisible(10, 5)` should be `True`.
* `isDivisible(10, 3)` should be `False`.

````python
def isDivisible(x, y):
    return x % y == 0

# Unit Tests (Scaffolding)
# Test Class 1: True case
assert isDivisible(20, 4) == True, "Error: 20 should be divisible by 4"

# Test Class 2: False case
assert isDivisible(20, 3) == False, "Error: 20 should not be divisible by 3"

print("All tests passed!")
````

# 7.9. Glossary

This glossary summarizes the foundational concepts of Boolean logic and conditional execution in Python.

### Core Logic Terms
* **boolean value**: There are exactly two: `True` and `False`. They belong to the type `bool`.
* **boolean expression**: An expression that evaluates to a boolean value.
* **boolean function**: A function that returns a boolean value.
* **comparison operator**: Operators used to compare values (`==`, `!=`, `>`, `<`, `>=`, `<=`).
* **logical operator**: Operators used to combine boolean expressions (`and`, `or`, `not`).
* **modulus operator**: Denoted by `%`, it returns the remainder of a division, often used in conditions to check for divisibility.



### Structural Terms
* **block**: A group of consecutive statements with the same indentation.
* **body**: The block of statements following a header in a compound statement.
* **condition**: The boolean expression in a conditional statement that determines the path taken.
* **conditional statement**: A statement that controls the flow of execution (using `if`, `elif`, `else`).

### Execution Flow Terms
* **branch**: One of the possible paths of execution in a conditional.
* **chained conditional**: A series of conditional branches using `if`, `elif`, and `else` to handle more than two possible paths.
* **nesting**: Placing one program structure inside another (e.g., an `if` statement inside the branch of another `if`).

````python
# Demonstrating Glossary Terms in Action

def is_even(n):
    """A boolean function using the modulus operator."""
    return n % 2 == 0

def check_number(x):
    # A conditional statement
    if x > 0: # The 'condition'
        # This is a block/body
        print("Positive")

        # NESTING: an if inside an if
        if is_even(x):
            print("and even")

    # CHAINED CONDITIONAL using elif
    elif x < 0:
        print("Negative")

    # The final branch
    else:
        print("Zero")

# Example calls
check_number(10)
check_number(-5)
````
